Python provides a powerful module called `itertools` that can generate permutations automatically using the `permutations` function. Let's see how we can use `itertools` for permutations.


```python
from itertools import permutations
```

####Question 1

How can we find all the possible ways to arrange the letters $\text A$, $\text B$, and $\text C$?


```python
items = ['A', 'B', 'C']
perms = list(permutations(items))

print("Permutations:", perms)
print("Number of permutations:", len(perms))
```

#### Question 2

How could we pick $2$ letters out of $\text A$, $\text B$, and $\text C$?

```python
items = ['A', 'B', 'C']
k = 2
perms_2 = list(permutations(items, k))

print("Permutations:", perms_2)
print("Number of permutations:", len(perms_2))
```

#### Question 3

How could we pick $3$ letters out of $\text A$, $\text B$, $\text C$ and $\text D$?

```python
items = ['A', 'B', 'C', 'D']
k = 3
perms_3 = list(permutations(items, k))

print("Permutations:", perms_3)
print("Number of permutations:", len(perms_3))
```

```python
from itertools import permutations

# Define the list of elements
elements = ['a', 'b', 'c', 'd']
items = 2

# Generate all permutations of 2 elements
perm = list(permutations(elements, items))

# Print the permutations
print(perm)
print("Number of ways to arrange 2 elements out of 4:", len(perm))
```
