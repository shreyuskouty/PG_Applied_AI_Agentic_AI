#### Question

Consider the word $\text {INFORMATION}$.  In how many unique ways can you rearrange the letters of this word?

---

When dealing with the number of unique arrangements of a word with repeating letters, we use a variation of the permutation formula. If we have $n$ objects where there are $n_1$ identical objects of type $1$, $n_2$ identical objects of type $2$, ..., and $n_k$ identical objects of type $k$, then the number of distinct permutations is given by:

$$\frac{n!}{n_1! \times n_2! \times \dots \times n_k!}$$


In the case of $\text{INFORMATION}$, total letters are $11$. Of all the letters, only $\text I$, $\text N$, and $\text O$ are repeated twice.

So, $\text{ }n=11$,  $\text{ }n_{\text{I}}=2$,  $\text{ }n_{\text{N}}=2$,$\text{ }$  and  $\text{ }n_{\text{O}}=2$

The number of unique permutations is
$$ \frac{n!}{n_{\text{I}}! \times n_{\text{N}}! \times n_{\text{O}}!} = \frac{11!}{2! \times 2! \times 2!}= \frac{39916800}{2 \times 2 \times 2} = \frac{39916800}{8} = 4989600$$

```python
# Count the occurrences of each letter
def count_and_permute(word):
    letter_counts = {}
    for letter in word:
        if letter in letter_counts:
            letter_counts[letter] += 1
        else:
            letter_counts[letter] = 1

    total_letters = len(word)

    # Define a function for factorial
    def factorial(n):
        if n == 0 or n == 1:
            return 1
        return n * factorial(n - 1)

    # Calculate the denominator (product of factorials of repeated letter counts)
    denominator = 1
    for count in letter_counts.values():
        denominator *= factorial(count)

    # Calculate the number of unique permutations
    unique_permutations = factorial(total_letters) / denominator

    return unique_permutations

count_and_permute("INFORMATION")
```
