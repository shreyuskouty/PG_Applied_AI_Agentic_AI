You are leading a committee that is organising an inter-school talent show. Students from different schools are participating in various activities such as singing, dance, drama, and art. You need to decide their performance orders and form groups for different activities.

#### Question 1

Out of $10$ volunteers, you need to select $4$ to help backstage. In how many ways can select $4$ volunteers?

---

The total number of ways to arrange $10$ volunteers is $10!$. If we select $4$ volunteers, there are $4!$ ways to arrange those $4$ volunteers within the selected group. Similarly, there are $(10-4)!$ ways to arrange the remaining $6$ volunteers.

To find the number of combinations, we divide the total permutations of $10$ volunteers by the number of permutations of the chosen group and the number of permutations of the remaining group.

The number of combinations of choosing $k$ items from a set of $n$ items is given by $$\mathrm{C}(n, k) = \binom{n}{k} = \frac{n!}{k!(n-k)!}$$

For selecting $4$ volunteers from $10$, $$\mathrm{C}(10, 4) = \binom{10}{4} = \frac{10!}{4!(10-4)!} = \frac{10!}{4!6!}= 210$$

```python
def combination(n, r):
    # Define a function to calculate factorial
    def factorial(n):
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result

    # Calculate combinations using the nested factorial function
    return factorial(n) / (factorial(r) * factorial(n - r))

print('Number of ways to select 4 volunteers out of 10:', combination(10, 4))
```

#### Question 2

You have to create a judge panel from $8$ teachers. The panel needs $2$ music experts, $2$ drama experts, and $1$ sports teacher. If $3$ of them are music teachers, $3$ drama teachers, and $2$ sports teachers, in how many ways can you form the panel?

---


This problem involves multiple independent selections from different groups. To find the total number of ways to form the panel, we multiply the number of ways to make each individual selection.

- Ways to choose $2$ music experts from $3$: $\mathrm{C}(3, 2)$

- Ways to choose $2$ drama experts from $3$: $\mathrm{C}(3, 2)$

- Ways to choose $1$ sports teacher from $2$: $\mathrm{C}(2, 1)$

$$\mathrm{C}(3, 2) \times \mathrm{C}(3, 2) \times \mathrm{C}(2, 1) = \binom{3}{2} \times \binom{3}{2} \times \binom{2}{1} = \frac{3!}{2!(3-2)!} \times \frac{3!}{2!(3-2)!} \times \frac{2!}{1!(2-1)!} = 18$$

```python
print('Number of ways to form the panel:', combination(3, 2) * combination(3, 2) * combination(2, 1))
```

#### Question 3

In how many ways can you split $6$ students into two teams of $3$ each?

---

First, consider selecting $3$ students for the first team. The number of ways to do this is $\mathrm{C}(6, 3)$. Let's say you select $\text{\{S1, S2, S3\}}$ for the first team. The other team is automatically $\text{\{S4, S5, S6\}}$.

However, if you had initially selected $\text{\{S4, S5, S6\}}$ for the first team, the other team would be $\text{\{S1, S2, S3\}}$.

Since the two teams are not labelled, the pair of teams $(\text{\{S1, S2, S3\}}, \text{\{S4, S5, S6\}})$ is the same as the pair $(\text{\{S4, S5, S6\}}, \text{\{S1, S2, S3\}})$.

We have counted each unique pairing of teams twice. To correct for this overcounting, we divide the number of ways to choose the first team by $2$.

$$\frac{\mathrm{C}(6, 3)}{2} = \frac{\binom{6}{3}}{2} = \frac{\frac{6!}{3!(6-3)!}}{2} = \frac{\frac{6!}{3!3!}}{2} = 10$$

```python
combination(6, 3) / 2
```
