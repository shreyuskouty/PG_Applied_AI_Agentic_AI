You are leading a committee that is organising an inter-school talent show. Students from different schools are participating in various activities such as singing, dance, drama, and art. You need to decide their performance orders and form groups for different activities.

#### Question 1

You have $3$ participants, $\text A $, $\text B$, and $\text C$ selected for final dance round. In how many ways can they be arranged to perform one after the another?

---

There are $6$ possible arrangements, which are:

- $\text {ABC}$

- $\text {ACB}$

- $\text {BAC}$

- $\text {BCA}$

- $\text {CAB}$

- $\text {CBA}$


Mathematically, the number of ways to arrange $n$ distinct objects is given by $n!$ where

$n! = 1 \times 2 \times ... \times (n-1) \times n$ for $n \neq 0$ and

$0! = 1$ (i.e., when $n = 0$)


Therefore for $3$ participants, the number of arrangements is $3! = 1 \times 2 \times 3 = 6$



```python
# Define a factorial function
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

factorial(3)
```

#### Question 2

Only $3$ out of the $5$ singers can perform in the final segment. In how many ways can you select and arrange them?

---




The number of permutations of selecting $k$ objects from a set of $n$ distinct objects, denoted as $\mathrm{P}(n, k)$ or $^n\mathrm{P}_k$, is given by $$\mathrm{P}(n, k) = \frac{n!}{(n-k)!} $$




In this case, we have $n = 5$ singers and we want to select and arrange $k = 3$ of them


$$ \mathrm{P}(5, 3) = \frac{5!}{(5-3)!} = \frac{5!}{2!}$$


$$ \mathrm{P}(5, 3) = \frac{1 \times 2 \times 3 \times 4 \times 5}{1 \times 2} = \frac{120}{2} = 60$$

```python
def permutation(n, k):
    return factorial(n) / factorial(n - k)

permutation(5, 3)
```
